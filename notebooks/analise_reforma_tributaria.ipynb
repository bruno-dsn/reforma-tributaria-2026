Este notebook confere as bases usadas no painel. Ele separa fatos legais, dados oficiais e cenários hipotéticos.

## Perguntas da análise

1. Como a transição evolui de 2026 a 2033?
2. Quais tratamentos de alíquota aparecem na LC 214?
3. Como os nove grupos do IPCA participam da cesta de referência em julho de 2026?
4. Como uma alíquota hipotética reage às reduções de 30%, 60% e 100%?

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
cronograma = pd.read_csv(ROOT / 'data' / 'cronograma_transicao.csv')
tratamentos = pd.read_csv(ROOT / 'data' / 'tratamentos_legais.csv')
ipca = pd.read_csv(ROOT / 'data' / 'ipca_grupos_2026_07.csv')
```

```python
cronograma[['ano', 'fase', 'cbs', 'ibs', 'icms_iss']]
```

```python
tratamentos.groupby(['tratamento', 'reducao_pct']).size().rename('categorias')
```

```python
assert abs(ipca['peso_mensal_pct'].sum() - 100) < 0.01
ipca.sort_values('peso_mensal_pct', ascending=False)[['grupo', 'peso_mensal_pct']]
```

```python
aliquota_hipotetica = 28
pd.DataFrame({'redução (%)': [0, 30, 60, 100]}).assign(
    **{'alíquota efetiva (%)': lambda df: aliquota_hipotetica * (1 - df['redução (%)'] / 100)}
)
```
